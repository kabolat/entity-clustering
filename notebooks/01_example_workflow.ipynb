# Entity-clustering example

This small, deterministic workflow runs on the bundled two-system dataset.

```python
from pathlib import Path

from entity_clustering.config import EncoderDefaults
from entity_clustering.data import load_daily_profiles
from entity_clustering.pipeline import EntityClusterer

data = load_daily_profiles(
    Path('../data/example/X_daily_15min_example.csv'),
    Path('../data/example/metadata.csv'),
    metadata_id_template='ID{:03d}',
)
model = EntityClusterer(
    number_topics=2, wording_granularity=8, number_clusters=1,
    number_lower_dims=96, distance_measure='bhattacharyya', linkage='average',
    random_state=2112, encoder=EncoderDefaults(max_iter=10, evaluate_every=0),
)
result = model.fit_predict(data.values)
assert result.labels.tolist() == [0, 0]
result.gamma
```
